## **[실습 3-1] 계산기 APP**

나눗셈시에는 만약 0으로 나누면 "Error : Division by Zero" 라는 문구가 뜨도록 만들어보자.

```python
# code here
import gradio as gr

def division(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Division by Zero"


demo = gr.Interface(
    fn = division,                                                              
    inputs = [                        
        gr.Number(label="변수1"),                                        
        gr.Number(label="변수2"),
    ],
    outputs= gr.Textbox(label="출력")
)

demo.launch()
```

## **[실습 3-2] BMI APP**

키는 cm, 몸무게는 kg로 입력받고, BMI 지수와 현재 상태를 표기하는 문자열까지 출력하도록 해보자.

image.png

```python
import gradio as gr

def BMI_Caculator(height, weight):
    height_m = height * 0.01
    bmi = weight / (height_m ** 2)

    if bmi < 18.5:
        status = "저체중"
    elif bmi < 23:
        status = "정상"
    elif bmi < 25:
        status = "과체중"
    elif bmi < 30:
        status = "경도비만"
    else:
        status = "고도비만"

    return status, round(bmi, 2)

demo = gr.Interface(
    fn=BMI_Caculator,
    inputs=[
        gr.Number(label="키 (cm)"),
        gr.Number(label="몸무게 (kg)"),
    ],
    outputs=[
        gr.Textbox(label="체중 정보"),
        gr.Textbox(label="BMI 지수"),
    ],
)

demo.launch()
```
