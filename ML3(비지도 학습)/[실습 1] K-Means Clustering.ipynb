## **1. 데이터 생성하기**

**Q1. sklearn.datasets 를 활용해서 3개의 군집 데이터를 생성합니다.**

```python
from sklearn.datasets import make_blobs

X, y = make_blobs(n_samples=200, centers=3, random_state=42)
```

**Q2. 생성된 데이터를 시각화해봅시다.**

```python
# code here
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(x = X[:,0], y =X[:,1], hue = y)
```

## **2. 모델링 1**

**Q1. K Means Clustering 비지도 학습 모델을 생성합니다.**

- n_clusters = 2로 먼저 설정해 봅시다.

```python
# code here
from sklearn.cluster import KMeans

model = KMeans(n_clusters = 2)
```

**Q2. 모델을 학습시킵니다.**

```python
# code here
model.fit(X)
```

**Q3. centroid의 위치를 데이터 위에 시각화해 봅시다.**

- model.cluster_centers_ : k개의 centroid의 좌표를 확인할 수 있습니다.
- model.labels_ : 각 데이터가 어떤 centroid(label)이 달렸는지를 확인할 수 있습니다.

```python
# code here
model.cluster_centers_
```

```python
model.labels_
```

```python
model.inertia_
```

```python
sns.scatterplot(x = X[:,0], y = X[:,1],  hue = model.labels_)
#centroid위치도 시각화
sns.scatterplot(x = model.cluster_centers_[:,0], 
                y = model.cluster_centers_[:,1], 
                marker="*", 
                color = "red",
                s = 200)
plt.show()
```

## **3. 모델링 2**

**Q1. 엘보우 메소드를 활용하여 최적의 K Value를 찾아 봅시다.**

```python
# code here
def elbow_method():
    inertia = []
    # k centroid만큼 KNN처럼 때려보고 intertia값을 비교해서 제일 최적화된 값을 우리의 k로 설정
    for k in range(1, 11):
        kmeans = KMeans(n_clusters = k)
        kmeans.fit(X)  # 학습
        inertia.append(kmeans.inertia_)  # k개의 그루븡로 clustering했을 때의 wcss를 일단 리스트에 순서대로 저장

    #시각화
    plt.figure(figsize=(5,3))
    sns.lineplot(x = range(1, 11), y = inertia, linestyle = "--", marker = "^")
    plt.xticks(range(1,11))
    plt.show()
```

```python
elbow_method()
```

**Q2. K Means Clustering 비지도 학습 모델을 생성합니다.**

- 엘보우 메소드에서 나온 n_clusters의 개수로 설정합니다.

```python
# code here
model = KMeans(
    n_clusters= 3,
    init = 'k-means++'  # random initalization trap 방지용 centroid 초기 배치를 위한 알고리즘
)
```

```python
model.fit(X)
```

**Q3. 데이터의 scatterplot을 그리고, 각 데이터가 어떤 centroid에 분배되었는지를 시각화합니다.**

- model.cluster_centers_ : k개의 centroid의 좌표를 확인할 수 있습니다.
- model.labels_ : 각 데이터가 어떤 centroid(label)이 달렸는지를 확인할 수 있습니다.

```python
# code here
sns.scatterplot(x = X[:,0], y = X[:,1],  hue = model.labels_)
#centroid위치도 시각화
sns.scatterplot(x = model.cluster_centers_[:,0], 
                y = model.cluster_centers_[:,1], 
                marker="*", 
                color = "red",
                s = 200)
plt.show()
```
