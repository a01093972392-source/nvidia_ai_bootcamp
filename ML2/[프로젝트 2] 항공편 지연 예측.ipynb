# **실습 개요**

여러분은 국내 항공사의 운항 데이터 분석팀에 합류한 데이터 사이언티스트입니다.<br>
항공편 지연은 승객 불편뿐 아니라 항공사의 평판과 비용에 직접적인 영향을 미칩니다.

출발 전에 알 수 있는 정보(날씨, 공항 혼잡도, 항공사, 시간대 등)만으로<br>
해당 항공편이 **15분 이상 지연될지(1) / 정시 출발할지(0)** 를 예측합니다.

## **데이터셋 설명**

| 컬럼명 | 설명 |
|---|---|
| flight_id | 항공편 고유 ID |
| month | 출발 월 (1~12) |
| day_of_week | 요일 (0=월요일, 6=일요일) |
| dep_hour | 출발 시각 (6~22시) |
| airline | 항공사 코드 (KAL, AAR, JJA, TWB, ESR) |
| dep_airport | 출발 공항 코드 (ICN, GMP, PUS 등) |
| arr_airport | 도착 공항 코드 |
| distance_km | 노선 비행 거리 (km) |
| weather_index | 기상 악화 지수 (0=맑음, 1=폭풍 수준) |
| dep_congestion | 출발 공항 혼잡도 (0~1) |
| arr_congestion | 도착 공항 혼잡도 (0~1) |
| airline_reliability | 항공사 정시 운항 신뢰도 (0~1, 높을수록 좋음) |
| is_rush_hour | 러시아워 여부 (7~9시, 18~21시) |
| is_weekend | 주말 여부 (토~일) |
| is_peak_season | 성수기 여부 (1월, 7~8월, 12월) |
| prev_delay_factor | 전날 해당 노선의 지연 누적 지수 (도미노 효과) |
| delay_minutes | 실제 지연 시간 (분, 음수=조기 출발) |
| is_delayed | **[target]** 15분 이상 지연 여부 (0=정시, 1=지연) |
