# **실습 개요**

여러분은 한 영화 리뷰 플랫폼(가상)의 데이터 분석팀에 합류했습니다.<br>
플랫폼에는 매일 수만 건의 관람객 리뷰가 올라오는데, 사람이 일일이 긍정/부정을 태깅하기에는 시간이 너무 오래 걸립니다.<br>
관람객이 남긴 리뷰 텍스트만으로 **긍정 리뷰인지 부정 리뷰인지 자동으로 분류**하는 감성 분석 모델을 만들어 리뷰 태깅 작업을 자동화해 달라고 합니다.

## **데이터셋 설명**

관람객이 남긴 영화 리뷰 텍스트와 감성 레이블 데이터입니다. train 데이터와 test 데이터가 이미 분리되어 제공됩니다.

**실제 네이버 영화 리뷰 데이터입니다.**

SOURCE : https://github.com/e9t/nsmc

| 컬럼명 | 설명 |
|---|---|
| id | 리뷰 고유 ID |
| document | 리뷰 텍스트 |
| label | **[target]** 감성 레이블 (0 = 부정, 1 = 긍정) |


## **데이터셋 URL**
train 데이터 : https://raw.githubusercontent.com/e9t/nsmc/master/ratings_train.txt

test 데이터 : https://raw.githubusercontent.com/e9t/nsmc/master/ratings_test.txt


## **1. 데이터 살펴보기**

데이터를 불러오고, 각 컬럼이 어떤 의미를 갖는지, 데이터의 전체적인 형태가 어떤지 살펴봅니다.

**Q1. 필요한 라이브러리를 불러오고, train/test 데이터를 각각 `train_df`, `test_df`에 저장하세요.**

- 이번 데이터 파일들을 tab (`\t`)으로 구분되어 있습니다. `read_csv()`에 `sep="\t"` 옵션을 추가하세요.

```python
import pandas as pd
train_df = pd.read_csv("https://raw.githubusercontent.com/e9t/nsmc/master/ratings_train.txt", sep='\t')
test_df = pd.read_csv("https://raw.githubusercontent.com/e9t/nsmc/master/ratings_test.txt", sep='\t')
```

```python
train_df.head()
```

## **2. 데이터 전처리하기**

### **2-1. 결측치 제거하기**

**Q1. 학습, 테스트 데이터에 결측치가 있는지 확인하세요.**

```python
# code here

train_df.isnull().sum()

test_df.isnull().sum()
```

**Q2. 확인된 결측치 행을 제거해 주세요.**

```python
# code here
# train_df 결측치 제거
train_df = train_df.dropna()
```

```python
# code here
# test_df 결측치 제거
test_df = test_df.dropna()
```

```python
#input  
X_train = train_df["document"]  
X_test = test_df["document"]
```

```python
# output
y_train = train_df["label"]
y_test = test_df["label"]
```

### **2-2. 데이터 벡터화하기**

**Q1. CountVectorizer를 활용해서 데이터를 벡터화합니다.**

등장한 단어를 0, 1로 존재함이 아닌, 등장 빈도를 셀 수 있도록 합니다.

```python
# code here
from sklearn.feature_extraction.text import CountVectorizer


cv_ml = CountVectorizer()

X_train_ml = cv_ml.fit_transform(X_train)
X_test_ml = cv_ml.transform(X_test)
```

## **3. 모델링 (MultinomialNB)**

**Q1. MultinomialNB 모델을 만들고 학습시키세요.**

```python
# code here

from sklearn.naive_bayes import MultinomialNB

model_ml = MultinomialNB()

model_ml.fit(X_train_ml, y_train)
```

**Q2. 학습된 모델로 X_test에 대한 예측을 수행합니다.**

```python
# code here
y_pred_ml = model_ml.predict(X_test_ml)   
```

**Q3. 모델의 성능을 평가합니다.**

```python
# code here
from sklearn.metrics import accuracy_score

accuracy_score(y_test, y_pred_ml)
```

## **4. 모델링 (BerNouliNB)**

**Q1. CountVectorizer를 활용해서 데이터를 벡터화합니다.**
- binary = True 옵션을 켜줍니다.

```python
# code here

vect_bn = CountVectorizer(binary=True)


X_train_bn = vect_bn.fit_transform(X_train)
X_test_bn = vect_bn.transform(X_test)
```

**Q2. BernoulliNB 모델을 만들고 학습시키세요.**

```python
# code here

from sklearn.naive_bayes import BernoulliNB

model_bn = BernoulliNB()

model_bn.fit(X_train_bn, y_train)
```

**Q3. 학습된 모델로 테스트 데이터에 대한 예측을 수행합니다.**

```python
# code here
y_pred_bn = model_bn.predict(X_test_bn)
```

**Q4. 모델의 성능을 평가합니다.**

```python
# code here
accuracy_score(y_test, y_pred_bn)
```
