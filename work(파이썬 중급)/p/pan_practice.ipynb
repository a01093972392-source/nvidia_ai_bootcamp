### 조건 결합
- and : & 
- or : |
- not : ~

주의: 모든 조건은 하나의 괄호로 묶어주는 습관

```python
# 조건 결합
# and or not
# 아까 나이가 30미만인 선수를 찾음
# 연봉이 4,000,000 이하인 데이터를 찾기

df[(df["나이"] < 30) & (df["급여"] <= 4000000)]
```

# 데이터 전처리


기본적으로 처리해야하는 값들
- 빈칸들(결측치)
- 중복 값
- 이상치(outlier)

```python
import pandas as pd
#url:https://raw.githubusercontent.com/dukalee/datasets/refs/heads/main/example_data2.csv
df = pd.read_csv("https://raw.githubusercontent.com/dukalee/datasets/refs/heads/main/example_data2.csv", encoding = 'utf8')
df
```

```python
#결측치: 칸이 어떤 값으로도 채워져있지 않은 "빈칸"을 표현
#DataFrame에서는 NaN이라는 표현을 쓴다.

df.isnull() # SQL
df.isna()   # numpy
```

```python
# 결측치 개수를 확인하기

df.isnull().sum()
```

```python
#1.삭제
df_clean = df.dropna() # NaN이 존재하는 행이 모두 삭제
```

```python
df_clean
```

```python
#결측치 처리법
#예를 들어 데이터의 양  = 100
#1. 만약 아주 소량의 데이터만 누락되어있다. -> 삭제
#   단, 이게 아주 중요한 정보 -> 최빈값 값은 것으로 채워준다.
#2. 너무 많은 데이터가 누락되어있다면 -> 특징을 빼야함
#   단, 이게 아주 중요한 정보 -> 데이터를 다시 모아야함.
```

```python
df2 = df.dropna(subset = ["나이", "몸무게"])
df2
```

```python
# fill 할 때는 일반적으로 넣는 값이 3개
# 1.평균값 : 솔직히 위험-> 평균으로 판단한다라는건 더 데이터를 많이 봐야함
# 2.중앙값 : 가장 안전한 방법 
# 3.최빈값 : 가장 많이 발생하는 값 -> 문자열 데이터의 경우 잘 사용
fill_value = df["급여"].mean()
fill_value

df2["급여"]= df["급여"].fillna(fill_value)
df2
```

```python
# 중복 데이터

df = df.drop(["index"], axis = 1)
df
```

```python
#df.duplicated(keep = False)  

# 중복 제거
df[~df.duplicated()]

df.drop_duplicates().reset_index(drop = True) 
```

```python
df.duplicated(subset = ['이름'])
```

# Pandas 고급

```python
# map, apply
# -> 한 열의 모든 정보에 대해서 특정 함수를 
#https://raw.githubusercontent.com/dukalee/datasets/refs/heads/main/example_data3.csv
import numpy as np
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/dukalee/datasets/refs/heads/main/example_data3.csv", encoding='utf8')
df
```

```python
# map
# 열의 데이터에 씌울 함수를 일단 하나 만들어야 한다
# 수학 성적의 점수를 200점 만점의 % 로 나타내고 싶다.
def convert(score):
    return round((score/200) * 100, 2)
    
```

```python
df_copy = df
df_copy["수학"] = df_copy["수학"].map(convert)
df_copy
```

```python
#1. split -> 원하는 정보들("1반/ 2반", "남/녀")
#2. apply -> filtering을 하는 작업
#3. combine -> DataFrame형태로 나타날 수 있도록 정리
```

```python
pd.Series(df.groupby("반"))
```

```python
pd.Series(df.groupby("반")["수학"])
```

```python
df.groupby("반")["수학"].mean()
```

```python
df.groupby("반")['수학'].max()
```

```python
df.groupby("반")[["수학", "국어"]].mean()
```

```python
df.groupby(["반","성별"])[["수학", "국어"]].mean()
```

```python
df.groupby(["반","성별"])[["수학", "국어"]].agg(["mean", "max"])
```
