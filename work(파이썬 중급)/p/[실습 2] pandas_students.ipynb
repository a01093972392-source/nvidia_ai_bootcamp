# **students.csv**


```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/dukalee/pandas_exercises/refs/heads/main/students.csv")
```

```python
df.head()
df
```

```python
ans = []
```

###**Q1. math_score가 가장 높은 학생은 누구인가?**

```python
# code here
q1 = df[df["math_score"] == df["math_score"].max()]["name"].values[0]
ans.append(q1)
```

###**Q2. english_score가 가장 낮은 학생은 누구인가?**

```python
# code here
q2 = df[df["english_score"] == df["english_score"].min()]["name"].values[0]
ans.append(q2)
```

###**Q3. 'New York'에 거주하는 학생은 총 몇 명인가?**

```python
# code here
q3 = len(df[df["city"] == "New York"])
ans.append(q3)
```

###**Q4. math_score의 평균은? (소수 둘째 자리 반올림)**

```python
# code here
q4 = df["math_score"].mean().round(1)
ans.append(q4)
```

###**Q5. 'Noah Davis' 학생의 city는?**

```python
# code here
q5 = df[df["name"] == "Noah Davis"]["city"].values[0]
ans.append(q5)
```

###**Q6. grade가 1인 학생은 몇 명인가?**

```python
# code here
q6 = len(df[df["grade"] == 1])
ans.append(q6)
```

###**Q7. total_score가 가장 높은 학생은 누구인가?**

```python
# code here
q7 = df[df["total_score"] == df["total_score"].max()]["name"].values[0]
ans.append(q7)
```

###**Q8. math_score가 80점 이상인 학생은 몇 명인가?**

```python
# code here
q8 = len(df[df["math_score"] >= 80])
ans.append(q8)
```

###**Q9. english_score의 최댓값은?**

```python
# code here
q9 = df["english_score"].max()
ans.append(q9)
```

###**Q10. 전체 학생 수는 몇 명인가?**

```python
# code here
q10 = len(df)
ans.append(q10)
```

## **제출**

```python
a = pd.DataFrame({
    "question_no" : [i+1 for i in range(10)],
    "ans" : ans
})
```
