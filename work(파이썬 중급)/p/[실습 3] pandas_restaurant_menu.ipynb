# **restaurant_menu.csv**

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/dukalee/pandas_exercises/refs/heads/main/restaurant_menu.csv")
df.head()
```

```python
ans = []
ans
```

### **Q1. price가 가장 높은 item은?**

```python
# code here

q1 = df[df["price"] == df["price"].max()]["item"].values[0]
ans.append(q1)
```

### **Q2. category가 'Main'인 item은 몇 개인가?**

```python
# code here
q2 = len(df[df["category"] == "Main"])
ans.append(q2)
```

###**Q3. orders_count가 가장 많은 item은?**

```python
# code here
q3 = df[df["orders_count"] == df["orders_count"].max()]["item"].values[0]
ans.append(q3)
```

###**Q4. calories가 500 이상인 item은 몇 개인가?**

```python
# code here
q4 = len(df[df["calories"] >= 500])
ans.append(q4)
```

###**Q5. category별 평균 price가 가장 높은 category는?**

```python
# code here
q5 = df.groupby("category")["price"].mean().idxmax()
ans.append(q5)
```

###**Q6. price 기준으로 정렬했을 때 3번째로 비싼 item은?**

```python
# code here
#df.loc[df["price"].nlargest(3).idxmin()]
df_new = df.sort_values("price", ascending=False)

q6 = df_new["item"].values[2]
ans.append(q6)
```

###**Q7. orders_count * price(매출 추정치)가 가장 큰 item은?**

```python
# code here
df["total"]= df["orders_count"] * df["price"]
q7 = df.sort_values("total", ascending=False)["item"].values[0]
ans.append(q7)
```

###**Q8. calories가 가장 낮은 item은?**

```python
# code here
q8 = df[df["calories"] == df["calories"].min()]["item"].values[0]
ans.append(q8)
```

###**Q9. category가 'Soup'인 item들의 평균 calories는? (소수 둘째 자리 반올림)**

```python
# code here
q9 = df[df["category"] == "Soup"]["calories"].mean().round(2)
ans.append(q9)
```

###**Q10. price 기준 하위 4개 item의 orders_count 평균은? (소수 둘째 자리 반올림)**

```python
# code here
d10 = df.sort_values("price").head(4)
q10 = d10["orders_count"].mean().round(2)
ans.append(q10)
```
