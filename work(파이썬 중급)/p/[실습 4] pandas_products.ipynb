# **products.csv**

```python
import pandas as pd
df = pd.read_csv("products.csv")
df
```

```python
ans = []
```

###**Q1. price가 가장 비싼 상품의 product_name은?**

```python
# code here
q1 = df.sort_values("price", ascending=False)["product_name"].values[0]
ans.append(q1)
```

### **Q2. category가 'Electronics'이면서 rating이 4.0 이상인 상품은 몇 개인가?**

```python
# code here
q2 = len(df[(df["category"] == "Electronics") & (df["rating"] >= 4.0)])
ans.append(q2)
```

### **Q3. stock이 가장 적은 상품의 product_name은?**

```python
# code here
q3 = df.sort_values("stock")["product_name"].values[0]
ans.append(q3)
```

### **Q4. category별 평균 price가 가장 높은 category는?**

```python
# code here
q4 = df.groupby("category")["price"].mean().idxmax()
ans.append(q4)
```

### **Q5. price가 100 이하인 상품 중 rating이 가장 높은 상품의 product_name은?**

```python
# code here
q5 = df[(df["rating"] == df[df["price"] <= 100]["rating"].sort_values(ascending=False).values[0])]["product_name"].values[0]
q5
ans.append(q5)
```

### **Q6. rating 기준으로 2번째로 낮은 상품의 product_name은?**

```python
# code here
q6 = df.sort_values("rating").head(2).tail(1)["product_name"].values[0]
q6
ans.append(q6)
```

### **Q7. stock * price 값이 가장 큰 상품의 product_name은?**

```python
# code here
df["total"] = df["stock"] * df["price"]
q7 = df.sort_values("total",ascending=False)["product_name"].values[0]
ans.append(q7)
```

### **Q8. category가 'Clothing'이거나 'Grocery'인 상품은 몇 개인가?**

```python
# code here
q8 = len(df[(df["category"] == "Clothing") | (df["category"] == "Grocery")])
ans.append(q8)
```

### **Q9. price가 상위 25%(0.75 분위수)를 초과하는 상품은 몇 개인가?**

```python
# code here
df.describe()
q9 = len(df[df["price"] > 187.5])
ans.append(q9)
```

### **Q10. category별 상품 개수가 가장 많은 category는?**

```python
# code here
q10 = df.groupby("category").count()["product_id"].idxmax()
ans.append(q10)
```
