# **[athletes.csv] 실습 문제**


```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/dukalee/pandas_exercises/refs/heads/main/athletes.csv")
df
```

```python
df.head()
```

```python
ans = []
```

###**Q1. medals가 가장 많은 선수는 누구인가?**

```python
# code here
q1 = df[(df["medals"] ==df["medals"].max())]["name"].values[0]
ans.append(q1)
```

### **Q2. country가 'South Korea'인 선수는 몇 명인가?**

```python
# code here
#len(df[df["country"] == "South Korea"])
q2 = (df["country"] == "South Korea").sum()
ans.append(q2)
```

###**Q3. medals가 0인 선수는 몇 명인가?**

```python
# code here
q3 = (df["medals"] == 0).sum()
ans.append(q3)
```

###**Q4. 'South Korea' 선수들의 medals 총합은?**

```python
# code here

dc = df[df["country"] == "South Korea"]
q4 = dc["medals"].sum()
ans.append(q4)
```

###**Q5. age가 가장 어린 선수는 누구인가?**

```python
# code here
q5 = df[df["age"] == df["age"].min()]["name"].values[0]
q5
ans.append(q5)
```

###**Q6. sport가 'Archery'인 선수의 age는?**

```python
# code here
q6 = df[df["sport"] == "Archery"]["age"].values[0]
ans.append(q6)
```

###**Q7. South Korea 선수들의 평균 나이는? (소수점 2자리까지)**

```python
# code here
q7 = df[df["country"] == "South Korea"]["age"].mean().round(2)
ans.append(q7)
```

###**Q8. medals가 2개 이상인 선수는 몇 명인가?**

```python
# code here
df[df["medals"] >= 2]
q8 = len(df[df["medals"] >= 2])
ans.append(q8)
```

### **Q9. age 기준 상위 3명 중 medals가 가장 많은 선수는 누구인가?**

```python
# code here
#df_age = df.loc[df["age"].sort_values(ascending = False).head(3).index]
#df_age[df_age["medals"] == df_age["medals"].max()]["name"].values[0]

new_df = df.sort_values("age", ascending=False)
q9 = new_df[new_df["medals"] == new_df["medals"].max()]["name"].values[0]
ans.append(q9)
```

###**Q10. 'South Korea'가 아닌 선수들의 평균 medals는? (소수 둘째 자리 반올림)**

```python
# code here
q10 = df[df["country"] != "South Korea"]["medals"].mean().round(2)
q10
ans.append(q10)
```
