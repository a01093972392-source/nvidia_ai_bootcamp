```python
#차원(dimension)
#0차원 데이터: scalar(단일값) -> 단일값을 가진 배열(list와 비슷)
#1차원 데이터: vector -> 리스트와 비슷하게 여러 값을 저장
#2차원 데이터: matrix -> 중첩 리스트와 비슷하다.
#3차원 데이터: tensor -> 딥러닝에서 배울것!
```

```python
#가장 많이 쓰게 될 명령어 -> numpy 배열을 만드는 명령어
#np.array()
li = [1, 2, 3]
print(type(li))
arr = np.array(li)
print(type(arr), arr)
```

```python
#scalar
scalar = np.array(5)
print(scalar)
print(type(scalar))
print(scalar.ndim)  #dimension:차원
print(scalar.shape) #shape:형태

# shape 표현: (n, ) -> n개의 요소가 있다.
```

```python
#scalar 끼리의 연산
a = np.array(3)
b = np.array(4)
print(a + b)
print(a * b)
```

```python
#vector: 1차원 배열(리스트)
vec = np.array([1,2,3])
print(vec)
print(type(vec))
print(vec.ndim, vec.shape)
```

```python
# 리스트의 동작을 모두 할 수 있다.
# 데이터 접근 -> indexing, slicing

print(vec[0])
print(vec[1:])
```

```python
# vector의 연산
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a + b)

c = np.array(1)
print(a + c)
```

```python
print(a @ b) #@ 내적곱
# @ 연산자
# 형태가 안맞으면 -> 내적곱
# 형태가 맞으면 -> 합성곱
# (1 x 4) + (2 X 5) + (3 X 6)
```

```python
# matrix : 2차원 배열 (vector 안에 vector)
# 행과 열로 이루어진 데이터

mat = np.array([
    [1, 2 ,3],
    [4, 5, 6]
])

print(type(mat), mat)
print(mat.ndim, mat.shape)
```

```python
# matrix도 indexing과 slicing
print(mat[0, 2])
print(mat[1, : 2])
```

```python
# matrix간의 연산
a = np.array([
    [1, 2 ,3],
    [4, 5, 6]
])

b = np.array([
    [1, 1 ,1],
    [2, 2, 2]
])

c = np.array([1, 1]) # shape 가 맞지 않으면 할 수 없다.
print(a + b)
print(a * b)
print(a + c)         # matrix와 한 행의 모양만 맞으면 할 수 있다
```

```python
# 행렬곱
# GPU가 잘하는 연산
# -> 딥러닝(신경망) 가장 잘하는 연산 -> 모든 것은 행렬곱으로 처리
# -> 인공 신경망의 핵심 기능

#(2, 3)
a = np.array([
    [1, 2, 3],
    [2, 3 ,4]
])

#(3, 2)
b = np.array([
    [1, 2],
    [2, 3],
    [3, 4]
])

print(a @ b)
```
