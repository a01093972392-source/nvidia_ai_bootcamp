## **산점도 그래프 기본 명령어**

| 명령어 / 옵션 | 의미 |
|---|---|
| `plt.scatter(x, y)` | x, y 값으로 산점도 그리기 |
| `s=값` 또는 `s=[값 리스트]` | 점의 크기 (리스트로 주면 점마다 다르게) |
| `alpha=값` | 점의 투명도 (0: 완전 투명 ~ 1: 완전 불투명) |
| `c='색상'` | 점의 색상 |
| `c=[그룹 배열]` + `cmap='coolwarm'` + `plt.colorbar()` | 값(그룹)에 따라 점 색을 다르게, 색상표 범례 표시 |
| `marker='모양'` | 점의 모양 |

### **Q7. 다른 열의 값으로 점 크기 다르게 하기**

출석률을 각각의 데이터의 크기로 사용해보자.


image.png

```python
# code here
plt.title("공부시간과 점수의 관계")
plt.xlabel("공부시간(시간)")
plt.ylabel("점수")
#s : size(각 마커의 크기를 설정할 수 있다)
#size를 기반으로 원래는 x vs y의 관계만 확인 -> 추가적으로 하나의 정보를 더 얻을 수 있게 만들 수 있다.
#alpha : 투명도(%) - 겹쳐있는 데이터를 한눈에 볼 수 있다 -> cluster (군집화를 확인)
plt.scatter(df["공부시간"], df["점수"], s = df["출석률"], alpha = 0.5, color = 'blue', marker = 'o')
plt.show()
```

### **Q8. 색상으로 데이터 표현하기**

- '합격 여부'라는 새로운 열을 만든다. (점수가 70 이상이면 1 아니면 0이다.)
- 이 값을 C로 전달하여 데이터의 색상을 표시한다.
- cmap은 "coolwarm"으로 색상을 구분한다.


image.png

```python
# 새로운 합격률을 만들기
# df["점수"] >= 70
#success = [1 if v == True else 0 for v in  df["점수"] >= 7]
s = np.array(df["점수"] >= 70)
df["합격"] = s
df
```

```python
# code here
plt.title("공부시간과 점수의 관계")
plt.xlabel("공부시간(시간)")
plt.ylabel("점수")
#s : size(각 마커의 크기를 설정할 수 있다)
#size를 기반으로 원래는 x vs y의 관계만 확인 -> 추가적으로 하나의 정보를 더 얻을 수 있게 만들 수 있다.
#alpha : 투명도(%) - 겹쳐있는 데이터를 한눈에 볼 수 있다 -> cluster (군집화를 확인)
plt.scatter(df["공부시간"], df["점수"], s= df["출석률"], alpha = 0.5, marker = 'o', c = df["합격"], cmap = 'coolwarm', label="합격률")
plt.colorbar(label = "합격여부(1=합격)")
plt.legend()
plt.show()
```

### **Q9. 성별 기준으로 분리하기.**

- 성별 기준으로 나눠 남학생은 파란 동그라미, 여학생은 빨간 별로 시각화하자.
- 라벨을 달고 범례도 보여주도록 한다.

image.png

```python
# code here
male = df[df["성별"] == "남"]
female = df[df["성별"] == "여"]

plt.title("공부시간과 점수의 관계")
plt.xlabel("공부시간(시간)")
plt.ylabel("점수")
#s : size(각 마커의 크기를 설정할 수 있다)
#size를 기반으로 원래는 x vs y의 관계만 확인 -> 추가적으로 하나의 정보를 더 얻을 수 있게 만들 수 있다.
#alpha : 투명도(%) - 겹쳐있는 데이터를 한눈에 볼 수 있다 -> cluster (군집화를 확인)
plt.scatter(male["공부시간"], male["점수"], alpha = 0.5, marker = 'o', label = "남")
plt.scatter(female["공부시간"], female["점수"], alpha = 0.5, marker = '*', label = "여")
plt.legend()
plt.show()
```

### **Q10. 조건부 색상 배열 만들기**

- 점수가 70 이상이면 green, 미만이면 red인 색상 배열을 만들고 적용한다.
- (HINT : np.where)



```python
colors = np.where(df['점수'] >= 70, 'green', 'red')
plt.title('공부시간과 점수')
plt.xlabel('공부시간(시간)')
plt.ylabel('점수')
plt.scatter(df['공부시간'], df['점수'], c=colors, label="합격")
plt.legend()
plt.show()
```
