## **막대그래프 기본 명령어**

| 명령어 / 옵션 | 의미 |
|---|---|
| `plt.bar(카테고리, 데이터)` | 세로 막대 그래프 |
| `color='색상'` 또는 `color=[색상 리스트]` | 막대 색상 (막대그래프는 `c`가 아니라 `color`) |
| `width=값` | 막대의 가로 폭 (기본값 0.8) |
| `plt.bar()`를 여러 번 호출 | 여러 막대 데이터를 한 그래프에 |
| `plt.barh(카테고리, 데이터)` | 가로 막대 그래프 |
| `x - width/2`, `x + width/2` | 막대를 살짝 좌우로 옮겨서 서로 겹치지 않게 배치 |
| `plt.xticks(위치, 이름)` | 숫자 위치에 카테고리 이름을 표시 |

### **Q7. 여러 막대를 겹쳐서 그리기**

- 지난달 판매량은 폭이 넓은(width=0.6) 회색 막대로 먼저 그린다.
- 그 위에 이번달 판매량을 폭이 좁은(width=0.3) 하늘색 막대로 겹쳐서 그린다.
- label과 legend 옵션을 활용하여 각각 그래프가 무엇을 표현하는지 나타낸다.

image.png

````python
# code here
plt.title("이번달 판매량")
plt.xlabel("이번달 VS 지난달 판매량")
plt.ylabel("판매량(잔)")
plt.bar(df["메뉴"], df["지난달판매량"], color = 'lightgray', width = 0.6, label = "지난달")
plt.bar(df["메뉴"], df["이번달판매량"], color = 'skyblue', width = 0.3, label = "이번달")
plt.legend()
plt.show()
````

### **Q8. 여러 막대를 나란히 그리기**

1. 카테고리의 기본 위치(0, 1, 2, 3)를 만든다.
```
x = np.arange(len(df))
```

2. 기본 width가 0.8이니, 한 막대의 크기를 0.4로 잡는다.
```
width = 0.4
```

4. 지난달 막대는 그 위치에서 width의 절반만큼 왼쪽으로(x - width/2),
    
    이번달 막대는 오른쪽으로(x + width/2) 옮겨서 그린다.

3. plt.xticks(x, df['메뉴'])로 숫자 위치 대신 메뉴 이름이 보이도록 바꾼다.


image.png

````python
# code here
x = np.arange(len(df))

width = 0.4

# position을 바꾸고 싶다면 bar(위치, x, y)..
plt.bar(x-(width /2), df["지난달판매량"], width = width)
plt.bar(x+(width /2), df["이번달판매량"], width = width)
````

### **Q9. 정렬 후 순위 비교하기**

가격이 낮은 순서로 정렬한 뒤, 가로 막대 그래프로 메뉴별 가격 순위를 비교해보자.


image.png

````python
# code here
df_sort = df.sort_values("가격")
plt.barh(df_sort["메뉴"], df_sort["가격"])
````

### **Q10. 조건부 색상 적용하기**

- '증가율' 이라는 새로운 열을 만든다. ((이번달 판매량 - 지난달 판매량) / 지난달 판매량)
- 품목별 증가율을 막대 그래프로 시각화한다.
- 성장률이 0 이상이면 초록색으로, 미만이면 빨간색으로 막대 그래프를 칠한다.

image.png

````python
# code here
df["증가율"] = ((df["이번달판매량"] - df["지난달판매량"]) / df["지난달판매량"])
plt.title("메뉴별 판매량 증가율")

plt.xlabel("메뉴")
plt.ylabel("증가율(%)")

color_list = ['green' , 'red']
for v in df["증가율"]:
    if v > 0:
        color_list.append("green")
    else:
        color_list.append("red")

#color_list = ["green" if v >= 0 else "red" for v in df["증가율"]]

plt.bar(df["메뉴"], df["증가율"], color = color_list)


````
