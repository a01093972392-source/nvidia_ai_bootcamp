## **선 그래프 기본 명령어**

| 명령어 / 옵션 | 의미 |
|---|---|
| `plt.plot(x, y)` | x, y 데이터로 선 그래프 그리기 |
| `plt.title('제목')` | 그래프 제목 |
| `plt.xlabel('이름')`, `plt.ylabel('이름')` | x축, y축 이름 |
| `c='색상'` | 선의 색 (문자열, rgb, hexcode) |
| `ls='--'`, `ls=':'` | 선 스타일 (`-` 실선(기본), `--` 파선, `:` 점선) |
| `marker='o'/'s'/'^'/'d'/'*'` | 실제 데이터 지점 표시 |
| `label='이름'` + `plt.legend()` | 범례에 표시될 이름 지정 |
| `plt.plot()`을 여러 번 호출 | 여러 선(데이터)을 한 그래프에 함께 그리기 |

### **Q7. 여러 선을 한 번에 그리기**

- 제목을 '저점별 월 매출'로 변경한다.
- 강남점은 파란 실선 + 동그라미 마커로 그린다.
- 홍대점은 빨간 파선 + 세모 마커로 그린다.
- 건대점은 초록 점선 + 네모 마커로 그린다.



image.png

```python
# code here
plt.title("강남점 월별 매출")
#x축 라벨
plt.xlabel("월")
#y축 라벨
plt.ylabel("매출(만원)")
# color(c) = 영문으로 원하는 색상을 넣어주면 된다.
# linestyle = ':' 점선, '-' 실선(default), '--' 파선
# marker = 'o' 원, '^' 삼각형, '*' 별, 's' 사각형
plt.plot(df["월"], df["강남점"], c = 'b', linestyle ='-', marker = 'o', label = "강남점")
plt.plot(df["월"], df["홍대점"], c = 'r', linestyle ='--', marker = '^', label = "홍대점")
plt.plot(df["월"], df["건대점"], c = 'g', linestyle =':', marker = 's', label = "건대점")
plt.legend() # 범례 표기
plt.show()
```

### **Q8. 종합 연습 — pandas로 새 열을 만들어서 함께 그리기**

- "강남점_누적'이라는 새로운 열을 만든다. (HINT : cumsum()을 활용한다.)
- 강남점과 강남점_누적의 두 선을 함께 그린다.
- 제목은 "강남점 매출 vs 누적매출"로 한다.
- 선의 표현은 자유롭게 한다.

image.png

```python
# code here
plt.title("강남점 매출 VS 누적매출")
# #x축 라벨
plt.xlabel("월")
# #y축 라벨
plt.ylabel("매출(만원)")
# color(c) = 영문으로 원하는 색상을 넣어주면 된다.
# linestyle = ':' 점선, '-' 실선(default), '--' 파선
# marker = 'o' 원, '^' 삼각형, '*' 별, 's' 사각형
df["강남점 누적 매출"]= df["강남점"].cumsum()

plt.plot(df["월"], df["강남점"], c = 'b', linestyle ='-', marker = 'o', label = "강남점 매출")
plt.plot(df["월"], df["강남점 누적 매출"], c = 'orange',  marker = 's', label = "강남점 누적 매출")
plt.legend() # 범례 표기
plt.show()
```
