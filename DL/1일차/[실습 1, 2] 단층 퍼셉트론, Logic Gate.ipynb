- numpy array로 준비한다.

```python
# 주어진 데이터 (X1=10, X2=4, bias=1)
import numpy as np
X = np.array([[10, 4]])
bias = np.array([1]) 
y = np.array([1])   # 결과
```

**Q2. 단층 퍼셉트론을 구현한다.**

머신 러닝의 모델을 만드는것과 비슷한 구성으로 만들어보자.

**__init__**
- self.w : np.zeros로 0으로 초기화한다. (입력 데이터의 개수와 동일하게 준비한다.)
- self.bias : 0으로 초기화한다.

**step function**
- threshold를 50으로 잡고, theshold 미만이면 0, 아니면 1을 반환한다.

**predict**
- 선형방적식으로 값을 구한다. (wx + b)
- 그리고 해당 값을 step function에 넣은 결과를 반환한다.

**fit**
- epoch번동안 아래 과정을 반복한다.
- X의 데이터를 순서대로 하나씩 넣고, predict를 해서 예측 결과를 확인한다.
- 실제 결과와 예측 결과의 차이를 계산한다.
- 그리고 각 가중치와 절편을 다음과 같이 업데이트한다.
  weight += learning rate * (실제결과와 예측 결과의 차이) * i번째 데이터의 값

각 epoch마다 가중치와 절편이 어떻게 바뀌었는지 출력한다.

```python
# code here

# bias 처리에 대해서 
# 1. 우리가 실제로 학습시킬 때 반드시 bias를 추가할 필요는 없다.
# [x1, x2, 1] -> bias(절편을 추가한다)
# [x1, x2] -> bias X (절편을 추가하지 않는다.) 


# 2. bias에 대해서 가중치를 관린하는것 -> 2가지로 나뉜다.
# - bias에 대해서는 가중치를 곱하지 ㅇ낳는다.
# [x1, x2, 1] -> x1w1 + x2w2 + 1- > 이 층이 활성화가 되었다.
# - 지금처럼 bias에도 weight를 설정해줘서 하나의 입력처럼 처리하는 방법
# [x1, x2, 1] -> x1w1 + x2w2 + 1b
class Perceptron:
    # 각 입력 데이터에 대한 가중치를 설정하기 위해서, 입력데이터의 몇개의 '특징'이 있는지를 알아야 함.
    def __init__(self, dim):
        self.w = np.zeros(dim)
        self.b = 0
        
    def step_function(self, z):
        threshold = 50
        if z >= 50:
            return 1
        return 0

    def predict(self, X):
        # 새로운 입력 데이터에 대해서 결과를 내는 것 -> 선형함수 + step
        y_pred = X[0] * self.w[0]  + X[1] * self.w[1] + self.b  # y = mx + b
        return self.step_function(y_pred)  # 0 또는 1로 최종결과


    def fit(self, X, y, bias, lr, epochs):
        # fit -> 원래는 X, y 지금은 X, y, bias 
        # 학습률과 epoch은 모두 hyperparameter이니, 우리가 어느순간에는 정의/전달을 해줘야 한다.
        # 1. epoch수 만큼 우리는 update를 반복
        for epoch in range(epochs):
            # 2. 각 epoch별 '하나의' 데이터셋을 기반으로 에측과 업데이트
            for i in range(len(X)):
                # 3. 예측 (내 전체 데이터셋의 i번째 데이터에 대한 예측)
                y_pred = self.predict(X[i])
                # 4. 손실을 구하기 (MSE/CE ..)
                error = y[i] - y_pred
                # --- 여기서 손실을 기반으로 gradient(경사하강법)
                # 5. 이 에러를 기반으로 가중치를 update
                self.w[0] += lr * error * X[i][0]   # 10이라는 첫번째 값에 대해서 w1을 업데이트
                self.w[1] += lr * error * X[i][1]
                self.b += error * lr * bias

            # epoch별 가중치의 변화를 한번 출력
            print(f"epoch {epoch+1}")
            print(f"w1 : {self.w[0]}")
            print(f"w2 : {self.w[1]}")
            print(f"b : {self.b}")
 
    
```

**Q3. Learning rate = 0.1, epoch = 5로 학습시킨 후, 각 가중치의 값을 찾아라.**
- 제출을 위해 각 가중치의 값을 기록해둔다.

```python
# code here
p = Perceptron(2)  # [10, 4] 처럼 하나의 데이터셋은 2개의 특징이 있다.
p.fit(X, bias, y, 0.1, 5)
```

## **제출하기**

- 컬럼명은 1, 2, 3, index는 w1, w2, bias라는 3 x 3 크기의 DataFrame을 만든다.
- 각 컬럼에 Q1, Q2, Q3에서 찾은 w1, w2, bias 가중치 값을 순서대로 기입한다.

```python
# code here
```

```python
import numpy as np

class Perceptron:
    def __init__(self, dim):
        # 입력 특성(dim) 개수만큼 가중치 배열 0으로 초기화
        self.w = np.zeros(dim)
        self.b = 0.0
        
    def step_function(self, z):
        # z가 0 이상이면 1, 아니면 0 반환 (배열 및 단일값 모두 지원)
        return np.where(z >= 0, 1, 0)

    def predict(self, X):
        # 행렬 곱(np.dot)을 이용해 선형 결합 계산 (y = Xw + b)
        z = np.dot(X, self.w) + self.b
        return self.step_function(z)

    def fit(self, X, y, lr=0.1, epochs=10):
        for epoch in range(epochs):
            for i in range(len(X)):
                # i번째 데이터 샘플 예측
                y_pred = self.predict(X[i])
                
                # 오차 계산
                error = y[i] - y_pred
                
                # 가중치 및 편향 업데이트
                self.w += lr * error * X[i]
                self.b += lr * error
                
            print(f"epoch {epoch+1} -> w1: {self.w[0]:.2f}, w2: {self.w[1]:.2f}, b: {self.b:.2f}")
```
